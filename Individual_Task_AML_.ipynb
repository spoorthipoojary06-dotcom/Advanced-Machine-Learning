```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (accuracy_score,precision_score,recall_score,f1_score,confusion_matrix,classification_report)
```

```python
file_path = ("/content/Diabetes_prediction.csv")
df = pd.read_csv(file_path)
```

```python
print("\nFirst 5 rows of the dataset:")
display(df.head())
print("\nShape of the dataset:")
print(df.shape)
print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("\nDataset Information:")
df.info()
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df.columns = df.columns.str.strip()
print("\nCleaned Column Names:")
print(df.columns.tolist())
```

```python
X = df.drop("Diagnosis", axis=1)
y = df["Diagnosis"]
print("\nFeatures:")
print(X.head())
print("\nTarget:")
print(y.head())
print("\nNumber of features:", X.shape[1])
print("Number of samples:", X.shape[0])
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)
print("\nTraining data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
decision_tree = DecisionTreeClassifier(
    random_state=42
)
naive_bayes = GaussianNB()
knn = KNeighborsClassifier(
    n_neighbors=5
)
```

```python
decision_tree.fit(
    X_train_scaled,
    y_train
)

naive_bayes.fit(
    X_train_scaled,
    y_train
)

knn.fit(
    X_train_scaled,
    y_train
)
```

```python
y_pred_dt = decision_tree.predict(X_test_scaled)
y_pred_nb = naive_bayes.predict(X_test_scaled)
y_pred_knn = knn.predict(X_test_scaled)
```

```python
dt_accuracy = accuracy_score(
    y_test,
    y_pred_dt
)

dt_precision = precision_score(
    y_test,
    y_pred_dt,
    average="weighted"
)

dt_recall = recall_score(
    y_test,
    y_pred_dt,
    average="weighted"
)

dt_f1 = f1_score(
    y_test,
    y_pred_dt,
    average="weighted"
)
```

```python
nb_accuracy = accuracy_score(
    y_test,
    y_pred_nb
)

nb_precision = precision_score(
    y_test,
    y_pred_nb,
    average="weighted"
)

nb_recall = recall_score(
    y_test,
    y_pred_nb,
    average="weighted"
)

nb_f1 = f1_score(
    y_test,
    y_pred_nb,
    average="weighted"
)
```

```python
knn_accuracy = accuracy_score(
    y_test,
    y_pred_knn
)

knn_precision = precision_score(
    y_test,
    y_pred_knn,
    average="weighted"
)

knn_recall = recall_score(
    y_test,
    y_pred_knn,
    average="weighted"
)

knn_f1 = f1_score(
    y_test,
    y_pred_knn,
    average="weighted"
)
```

```python
results = pd.DataFrame({

    "Algorithm": [
        "Decision Tree",
        "Naive Bayes",
        "KNN"
    ],

    "Accuracy": [
        dt_accuracy,
        nb_accuracy,
        knn_accuracy
    ],

    "Precision": [
        dt_precision,
        nb_precision,
        knn_precision
    ],

    "Recall": [
        dt_recall,
        nb_recall,
        knn_recall
    ],

    "F1 Score": [
        dt_f1,
        nb_f1,
        knn_f1
    ]
})
print("\n================================================")
print("PERFORMANCE COMPARISON")
print("================================================")
display(results)
```

```python
results_percentage = results.copy()
results_percentage[
    ["Accuracy", "Precision", "Recall", "F1 Score"]
] = (
    results_percentage[
        ["Accuracy", "Precision", "Recall", "F1 Score"]
    ] * 100
).round(2)
print("\n================================================")
print("PERFORMANCE COMPARISON IN PERCENTAGE")
print("================================================")

display(results_percentage)
```

```python
print("\n================================================")
print("CLASSIFICATION REPORT - DECISION TREE")
print("================================================")
print(
    classification_report(
        y_test,
        y_pred_dt
    )
)
print("\n================================================")
print("CLASSIFICATION REPORT - NAIVE BAYES")
print("================================================")

print(
    classification_report(
        y_test,
        y_pred_nb
    )
)
print("\n================================================")
print("CLASSIFICATION REPORT - KNN")
print("================================================")

print(
    classification_report(
        y_test,
        y_pred_knn
    )
)
```

```python
cm_dt = confusion_matrix(
    y_test,
    y_pred_dt
)
plt.figure(figsize=(4, 4))
sns.heatmap(
    cm_dt,
    annot=True,
    fmt="d",
    cmap="Blues"
)
plt.title("Confusion Matrix - Decision Tree")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.show()
```

```python
cm_nb = confusion_matrix(
    y_test,
    y_pred_nb
)
plt.figure(figsize=(4, 4))
sns.heatmap(
    cm_nb,
    annot=True,
    fmt="d",
    cmap="Blues"
)
plt.title("Confusion Matrix - Naive Bayes")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.show()
```

```python
cm_knn = confusion_matrix(
    y_test,
    y_pred_knn
)
plt.figure(figsize=(4, 4))
sns.heatmap(
    cm_knn,
    annot=True,
    fmt="d",
    cmap="Blues"
)
plt.title("Confusion Matrix - KNN")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.show()
```

```python
results_percentage.set_index(
    "Algorithm"
)[
    ["Accuracy", "Precision", "Recall", "F1 Score"]
].plot(
    kind="bar",
    figsize=(6, 6)
)
plt.title(
    "Performance Comparison of ML Algorithms"
)
plt.xlabel("Algorithms")
plt.ylabel("Performance (%)")
plt.xticks(rotation=0)
plt.ylim(0, 100)
plt.legend(
    title="Evaluation Metrics"
)
plt.grid(
    axis="y"
)
plt.show()
```

```python
best_index = results["F1 Score"].idxmax()
best_algorithm = results.loc[
    best_index,
    "Algorithm"
]
best_accuracy = results.loc[
    best_index,
    "Accuracy"
]
best_precision = results.loc[
    best_index,
    "Precision"
]
best_recall = results.loc[
    best_index,
    "Recall"
]
best_f1 = results.loc[
    best_index,
    "F1 Score"
]
```

```python
print("\n")
print("================================================")
print("              FINAL RESULT")
print("================================================")
print(
    "\nBest Performing Algorithm:",
    best_algorithm
)
print(
    "Accuracy:",
    round(best_accuracy * 100, 2),
    "%"
)
print(
    "Precision:",
    round(best_precision * 100, 2),
    "%"
)
print(
    "Recall:",
    round(best_recall * 100, 2),
    "%"
)
print(
    "F1 Score:",
    round(best_f1 * 100, 2),
    "%"
)
print("\n================================================")
print("CONCLUSION")
print("================================================")

print(
    f"\nAmong Decision Tree, Naive Bayes and KNN, "
    f"{best_algorithm} performed the best based on "
    f"the highest F1 Score."
)
```
